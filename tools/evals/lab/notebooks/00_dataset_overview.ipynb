# Dataset 개요

평가 dataset split 하나를 살펴본다.

기본 split은 `dev`다. `held-out`은 `allow_held_out=True`가 필요하고 최종 확인용이다.

```python
from snapdone_eval_lab.datasets import load_dataset

dataset = load_dataset("sample-classification", split="dev")
dataset.frame.head()
```

```python
print(dataset.name, dataset.split, dataset.task, "행:", len(dataset.frame))
list(dataset.frame.columns)
```

```python
# label 분포. category 열은 분류 dataset에만 있다.
if "category" in dataset.frame:
    counts = dataset.frame["category"].value_counts()
    display(counts)
    counts.plot.bar(title=f"{dataset.name} {dataset.split}: cases per category", rot=0)
```
