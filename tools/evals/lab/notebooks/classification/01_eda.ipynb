# 분류 EDA

무엇이든 돌리기 전에 image-classification dataset을 살핀다.

실제 dataset이 생기면 `DATASET`을 `tools/evals/datasets` 아래 그 디렉터리 이름으로 바꾼다.

```python
from snapdone_eval_lab.datasets import load_dataset
from snapdone_eval_lab.plots import bar, show_images

DATASET = "sample-classification"
SPLIT = "dev"  # validation은 필요할 때. held-out은 allow_held_out=True가 필요하고 최종 확인용이다

dataset = load_dataset(DATASET, split=SPLIT)
frame = dataset.frame
print(dataset.name, dataset.split, dataset.task, "행:", len(frame))
frame.head()
```

```python
bar(frame["category"].value_counts(), "gold category")
frame["category"].value_counts()
```

```python
# 검토 상태 · 방법 · 모호함별 case 수
frame.groupby(["review", "method", "ambiguity"], dropna=False).size()
```

```python
# 의도와 case마다 기대하는 fact 수
frame.groupby(["intent", "fact_count"]).size()
```

```python
# 빈 값과 fixture 확인
display(frame.isna().sum())
frame.loc[frame["fixture_exists"] != True, ["id", "fixture_path", "fixture_exists"]]
```

```python
# 기본 걸러 보기: 조건을 바꿔 case를 자른다.
picked = frame[(frame["difficulty"] == "easy") & (frame["review"] == "reviewed")]
picked[["id", "category", "intent", "acceptable_actions", "forbidden_actions", "tags"]]
```

```python
# 고른 case의 앞쪽 fixture 이미지 미리보기
show_images(dataset, list(picked["id"][:6]))
```
